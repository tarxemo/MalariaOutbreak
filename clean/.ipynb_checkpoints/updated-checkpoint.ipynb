```python
import pandas as pd
import numpy as np

# -----------------------------
# Load datasets
# -----------------------------
humidity = pd.read_csv("combined_humidity.csv")
rainfall = pd.read_csv("combined_rainfall_data.csv")
temperature = pd.read_csv("combined_temperature_data.csv")

# -----------------------------
# 1. Clean Humidity
# -----------------------------
humidity['Date'] = pd.to_datetime(humidity['Date'])
humidity['Year'] = humidity['Date'].dt.year
humidity['Month'] = humidity['Date'].dt.month

# Expand each month to 4 weeks with same humidity value
humidity_weeks = []
for _, row in humidity.iterrows():
    for week in range(1, 5):  # Approximate 4 weeks per month
        humidity_weeks.append({
            "District": row["Region"].strip().upper(),  # Use as District
            "Island": row["Island"],
            "Humidity": row["Humidity"],
            "Year": row["Year"],
            "Month": row["Month"],
            "Week_Num": week
        })
humidity_df = pd.DataFrame(humidity_weeks)

# -----------------------------
# 2. Clean Rainfall and Temperature
# -----------------------------
# Convert month names to numbers
month_map = {
    "Jan": 1, "Feb": 2, "Mar": 3, "Apr": 4, "May": 5, "Jun": 6,
    "Jul": 7, "Aug": 8, "Sep": 9, "Oct": 10, "Nov": 11, "Dec": 12
}
rainfall['Month'] = rainfall['Month'].map(month_map)
temperature['Month'] = temperature['Month'].map(month_map)

# Normalize district names
rainfall['District'] = rainfall['District'].str.strip().str.upper()
temperature['District'] = temperature['District'].str.strip().str.upper()

# Convert "1st week", "2nd week", ... → week number
def week_to_num(week_str):
    mapping = {
        "1st week": 1, "2nd week": 2, "3rd week": 3,
        "4th week": 4, "5th week": 5
    }
    return mapping.get(str(week_str).strip().lower(), np.nan)

rainfall['Week_Num'] = rainfall['Week'].apply(week_to_num)
temperature['Week_Num'] = temperature['Week'].apply(week_to_num)

# -----------------------------
# 3. Merge Rainfall + Temperature
# -----------------------------
climate = pd.merge(
    rainfall,
    temperature,
    on=["District", "Station", "Year", "Month", "Week", "Week_Num"],
    how="outer"
)

# -----------------------------
# 4. Merge with Humidity
# -----------------------------
# Use District, Year, Month, Week_Num
final_climate = pd.merge(
    climate,
    humidity_df,
    on=["District", "Year", "Month", "Week_Num"],
    how="left"
)

# -----------------------------
# 5. Export Final Result
# -----------------------------
final_climate.to_csv("combined_climatic.csv", index=False)
print("✅ combined_climatic.csv has been saved.")
```
